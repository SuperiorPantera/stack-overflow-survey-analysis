```python
import pandas as pd

df = pd.read_csv("survey_results_public.csv", low_memory=False)
schema = pd.read_csv("survey_results_schema.csv")

df.head(15)
```

```python
schema.head(15)
```

```python
#Task_1
total_responds = df.shape[0]
total_responds
```

```python
#Task_2
questions = set(schema["qname"])

columns = set(df.columns)

valid_columns = list(questions.intersection(columns))

complete_responses = df[valid_columns].dropna().shape[0]
print(complete_responses)
```

```python
#Task_3
workexp = df["WorkExp"]

mean_val = workexp.mean()
median_val = workexp.median()
mode_val = workexp.mode()

print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val[0])
```

```python
#Task_4
remote_count = df[df["RemoteWork"] == "Remote"].shape[0]
print(remote_count)
```

```python
#Task_5
python_users = df["LanguageHaveWorkedWith"].str.contains("Python", na=False)

percentage = python_users.mean() * 100
print(f"Відсоток респондентів, які програмують на Python: {percentage:.2f}%")
```

```python
#Task_6
online_learners = df["LearnCode"].str.contains("Online Courses", case=False, na=False)

count_online = online_learners.sum()
print(count_online)
```

```python
#Task_7
python_df = df[df["LanguageHaveWorkedWith"].str.contains("Python", na=False)]

salary_stats = python_df.groupby("Country")["ConvertedCompYearly"].agg(Average_Compensation="mean", Median_Compensation="median").reset_index()
salary_stats
```

```python
#Task_8
top5 = df.sort_values(by="ConvertedCompYearly", ascending=False).head(5)

top5_education = top5[["ConvertedCompYearly", "EdLevel"]]
top5_education
```

```python
#Task_9
df["Python"] = df["LanguageHaveWorkedWith"].str.contains("Python", na=False)

age_python = df.groupby("Age")["Python"].mean().reset_index()
age_python["Python"] = age_python["Python"] * 100

age_python.columns = ["Age_category", "Python_percentage"]
age_python
```

```python
#Task_10
rich_percentile = df["ConvertedCompYearly"].quantile(0.75)

filtered = df[
    (df["ConvertedCompYearly"] >= rich_percentile) &
    (df["RemoteWork"] == "Remote")]

top_industries = filtered["Industry"].value_counts()
print(top_industries)
```
